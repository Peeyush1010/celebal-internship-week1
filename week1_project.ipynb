```python
# Unzip the file
import zipfile
import os

with zipfile.ZipFile('sales_data.zip', 'r') as zip_ref:
    zip_ref.extractall('sales_data')

print(" File unzipped successfully!")
print("Files extracted:")
for f in os.listdir('sales_data'):
    print(f"  - {f}")
```

```python
import pandas as pd
import numpy as np

# Load the main combined dataset
df = pd.read_csv('sales_data/Combined_dataset.csv')

print(" Dataset loaded successfully!")
print(f"   Rows    : {df.shape[0]:,}")
print(f"   Columns : {df.shape[1]}")
```

```python
# First and last rows
print("=== First 5 rows ===")
display(df.head())

print("\n=== Last 5 rows ===")
display(df.tail())
```

```python
# Shape and column names
print(f"Shape: {df.shape}  →  {df.shape[0]} rows × {df.shape[1]} columns")
print()
print("Column Names:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i:>2}. {col}")
```

```python
print("=== Data Types ===")
print(df.dtypes)
```

```python
df.info()
```

```python
display(df.describe())
```

```python
print("=== Top 15 Product Categories ===")
print(df['category'].value_counts().head(15))
print(f"\nTotal unique categories: {df['category'].nunique()}")
```

```python
print("=== Missing Values BEFORE Cleaning ===")
missing       = df.isnull().sum()
missing_pct   = (missing / len(df) * 100).round(1)
missing_table = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
display(missing_table[missing_table['Missing Count'] > 0])
print(f"\nTotal missing cells: {df.isnull().sum().sum()}")
```

```python
df['discount']            = df['discount'].fillna(0)
df['what_customers_said'] = df['what_customers_said'].fillna('No reviews')
df['seller_name']         = df['seller_name'].fillna('Unknown Seller')
df['seller_information']  = df['seller_information'].fillna('N/A')
df['variations']          = df['variations'].fillna('No variations')
df['videos']              = df['videos'].fillna('No video')

# Verify
print("=== Missing Values AFTER Cleaning ===")
print(df.isnull().sum())
print(f"\n Total missing cells remaining: {df.isnull().sum().sum()}")
```

```python
# Filter rows
high_rated = df[df['rating'] >= 4.0]
print(f"High-rated (≥4.0): {len(high_rated)} items")

big_discount = df[df['discount'] > 60]
print(f"Discount > 60%: {len(big_discount)} items")

# Select columns
display(df[['product_id', 'title', 'category', 'rating', 'initial_price', 'discount']].head(5))

# loc and iloc
display(df.loc[0:4, ['title', 'category', 'initial_price']])
display(df.iloc[0:5, 0:4])
```

```python
print(f"Before: {len(df)} rows | Duplicates: {df.duplicated().sum()}")
df = df.drop_duplicates().reset_index(drop=True)
print(f"After : {len(df)} rows")
```

```python
df['quantity']     = 1
df['total_amount'] = df['initial_price'] * df['quantity']

display(df[['title', 'initial_price', 'quantity', 'total_amount']].head(5))
```

```python
df.to_csv('Combined_dataset_cleaned.csv', index=False)
print(f" Saved! Shape: {df.shape}")

from google.colab import files
files.download('Combined_dataset_cleaned.csv')
```
